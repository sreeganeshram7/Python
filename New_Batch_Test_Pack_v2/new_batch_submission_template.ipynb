## Q1. Text Analysis Function (10 marks)

Write a function named `analyze_text(text)`.

The function must return a dictionary with these keys:
- `letters`: number of alphabetic characters
- `digits`: number of numeric characters
- `spaces`: number of spaces
- `vowels`: number of vowels (`a, e, i, o, u`) ignoring case

Ignore punctuation for the counts above.

### Test case
`analyze_text("Python Lab 2025!")`

### Expected output structure
`{"letters": 9, "digits": 4, "spaces": 2, "vowels": 2}`

````python
# Write your solution 
def analyze_text(text):
    result={"letters":0,"digits":0,"spaces":0,"vowels":0,"others":0}
    for ch in text:
        if ch.lower() in ['a','e','i','o','u']:
            result["vowels"]+=1
            result["letters"]+=1
        elif ch.isalpha():
            result["letters"]+=1
        elif ch.isdigit():
            result["digits"]+=1
        elif ch.isspace():
            result["spaces"]+=1
        else:
            result["others"]+=1
    return result
````

## Q2. Prime Numbers in a Range (10 marks)

Write a function named `get_primes(start, end)`.

Return a list of all prime numbers between `start` and `end` (both inclusive).

### Test case
`get_primes(20, 60)`

````python
# Write your solution below
def get_primes(start,end):
    r=[]
    for num in range(start,end+1):
        if num<2:
            continue
        for i in range(2,int(num**0.5)+1):
            if num%i==0:
                break
        else:
            r.append(num)
    return r
````

## Q3. Move a Selected Value to the End (10 marks)

Write a function named `move_value_to_end(num_list, value)`.

Move all occurrences of `value` to the end of the list while preserving the order of the remaining elements.

### Test case
`move_value_to_end([0, 2, 4, 5, 6, 2, 3], 2)`

### Expected output
`[0, 4, 5, 6, 3, 2, 2]`

````python
# Write your solution below
def move_value_to_end(num_list,value):
    r=[]
    c=0
    for num in num_list:
        if num==value:
            c+=1
        else:
            r.append(num)
    r.extend([value] * c)
    return r
````

## Q4. Dictionary and Set Report (10 marks)

Write a function named `dictionary_report(dict_a, dict_b)`.

The function must return a dictionary with these keys:
- `common_keys`: a sorted list of keys present in both dictionaries
- `same_value_pairs`: a sorted list of `(key, value)` tuples where the key exists in both dictionaries and the values are equal
- `combined_keys`: a set containing all keys from both dictionaries

Use these dictionaries for testing:
```python
num_1 = {'a': 5, 's': 7, 'x': 11, 'm': 12, 'o': 8}
num_2 = {'r': 12, 'x': 9, 'n': 8, 'm': 12, 'q': 10}
```

````python
# Write your solution below
def dictionary_report(dict_a, dict_b):
    common_keys = sorted(set(dict_a.keys()) & set(dict_b.keys()))
    unique_a = sorted(set(dict_a.keys()) - set(dict_b.keys()))
    unique_b = sorted(set(dict_b.keys()) - set(dict_a.keys()))

    return {
        "common_keys": common_keys,
        "unique_a": unique_a,
        "unique_b": unique_b
    }
````

````python
dict_a = {'a': 5, 's': 7, 'x': 11, 'm': 12, 'o': 8}
dict_b = {'r': 12, 'x': 9, 'n': 8, 'm': 12, 'q': 10}
dictionary_report(dict_a, dict_b)
````

## Q6. Convert Seconds into Days, Hours, Minutes, and Seconds (10 marks)

Write a function named `convert_seconds(total_seconds)`.

Return the result as a tuple in the order:
`(days, hours, minutes, seconds)`

### Test case
`convert_seconds(996452)`

### Expected output
`(11, 12, 47, 32)`

````python
# Write your solution below
def convert_seconds(total_seconds):
    days = total_seconds // 86400
    total_seconds %= 86400

    hours = total_seconds // 3600
    total_seconds %= 3600

    minutes = total_seconds // 60
    seconds = total_seconds % 60

    return (days, hours, minutes, seconds)
````

## Q7. Exception Handling with String Reversal (10 marks)

Write a function named `safe_reverse_even(text)`.

Rules:
- if the input is a string with even length, return the reversed string
- if the input is a string with odd length, return the original string
- if a non-string value is passed, handle the `TypeError` and return the message: `"Check the string."`
- for any other unexpected error, return: `"Something went wrong."`

### Demonstrate with these examples
- `safe_reverse_even("Python")`
- `safe_reverse_even("Coder")`
- `safe_reverse_even(123)`

````python
# Write your solution belodef safe_reverse_even(text):
def safe_reverse_even(text):
    try:
        if not isinstance(text, str):
            raise TypeError("Input must be a string")

        if len(text) % 2 != 0:
            raise ValueError("String length must be even")

        return text[::-1]

    except Exception as e:
        return str(e)
safe_reverse_even("Python")
````

## Q8. OOP and Inheritance (12 marks)

Create these classes:
- `Pet`
- `Dog(Pet)`
- `Cat(Pet)`

Requirements:
- the base class must store `name` and `species`
- create a method `display_info()` that prints the details clearly
- create these objects and display their details:
```python
Cat("Kitty", "Persian Cat")
Dog("Bruno", "German Shepherd")
```

````python
# Write your solution below
class Pet:
    def __init__(self, name, species):
        self.name = name
        self.species = species

    def display_info(self):
        return f"{self.name} is a {self.species}"


class Dog(Pet):
    def __init__(self, name):
        super().__init__(name, "Dog")


class Cat(Pet):
    def __init__(self, name):
        super().__init__(name, "Cat")
````

````python
cat = Cat("Kitty")
dog = Dog("Bruno")

print(cat.display_info())
print(dog.display_info())
````

## Q9. File Handling Summary (14 marks)

Write a function named `file_summary(path)` using the file:
`datasets/story.txt`

The function must return a dictionary with these keys:
- `line_count`: total number of lines
- `word_count`: total number of words
- `python_count`: number of times the word `python` appears, ignoring case and punctuation
- `longest_word`: the longest word in the file, converted to lowercase and without punctuation

Print the returned dictionary after calling the function.

````python
import string

def file_summary(path):
    line_count = 0
    word_count = 0
    python_count = 0
    longest_word = ""

    with open(path, 'r') as file:
        for line in file:
            line_count += 1

            # Remove punctuation and split words
            words = line.strip().split()
            clean_words = []

            for word in words:
                clean = word.strip(string.punctuation).lower()
                if clean:  # ignore empty
                    clean_words.append(clean)

            word_count += len(clean_words)

            for word in clean_words:
                # Count 'python'
                if word == "python":
                    python_count += 1

                # Find longest word
                if len(word) > len(longest_word):
                    longest_word = word

    return {
        "line_count": line_count,
        "word_count": word_count,
        "python_count": python_count,
        "longest_word": longest_word
    }


# Call function
result = file_summary("datasets/story.txt")
print(result)
````

## Q10. CSV Report using the Standard Library (14 marks)

Write a function named `build_student_report(input_path, output_path)` using the file:
`datasets/student_marks.csv`

Tasks:
1. Read the CSV file using the built-in `csv` module.
2. Calculate the average marks of each student across Math, Science, and English.
3. Assign grades using these rules:
   - `A` for average >= 85
   - `B` for average >= 70 and < 85
   - `C` for average >= 55 and < 70
   - `D` for average < 55
4. Create an output CSV file named `q10_student_report.csv` with these columns:
   - `Name`
   - `Average`
   - `Grade`
5. Round averages to 2 decimal places.
6. Sort the rows by `Average` in descending order before writing the output file.

At the end, print the rows you wrote to the CSV file.

````python
# Write you solution below
import csv

def build_student_report(input_path, output_path):
    students = []

    # Step 1: Read CSV
    with open(input_path, 'r') as infile:
        reader = csv.DictReader(infile)

        for row in reader:
            name = row['Name']
            math = float(row['Math'])
            science = float(row['Science'])
            english = float(row['English'])

            # Step 2: Calculate average
            avg = (math + science + english) / 3
            avg = round(avg, 2)

            # Step 3: Assign grade
            if avg >= 85:
                grade = 'A'
            elif avg >= 70:
                grade = 'B'
            elif avg >= 55:
                grade = 'C'
            else:
                grade = 'D'

            students.append({
                'Name': name,
                'Average': avg,
                'Grade': grade
            })

    # Step 4: Sort by Average (descending)
    students.sort(key=lambda x: x['Average'], reverse=True)

    # Step 5: Write output CSV
    with open(output_path, 'w', newline='') as outfile:
        fieldnames = ['Name', 'Average', 'Grade']
        writer = csv.DictWriter(outfile, fieldnames=fieldnames)

        writer.writeheader()
        writer.writerows(students)


# Call function
build_student_report("datasets/student_marks.csv", "q10_student_report.csv")
````
