```python
# Types of inheritance
# 1.Single Inheritance
# 2.Multiple Inheritance
# 3.Multilevel Inheritance
# 4.Hierarchical Inheritance
# 5.Hybrid Inheritance
```

```python
# Single Inheritance
# From one parent to one child to acquire the properties
class A:
    pass
class B:
    pass
```

```python
# Multilevel Inheritance
# From one parent to one child to grand child to acquire the properties
class A:
    pass
class B(A):
    pass
class C(B):
    pass
```

```python
# Multiple Inheritance
# From one child to multiple parents to acquire the properties
class A:
    def show(self):
        print(A)
class B:
    def show(self):
        print(B)
class C(A,B):
    pass
c=C()
c.show()
```

```python
print(C.mro())
```

```python
# MRO - Method Resoultion Order
# ---->it is the order which python looks for a method or attributes in a class hierarchy 
    #    especially when multiple inheritance is involved .It follows the C3 linearization algorithm
```

```python
# why MRO?
# to avoid the ambiguity in multiple inheritance 
# to decide which method to call first
# to supports super() correctly
# to solve the diamond problem
```

```python
# how python finds a method
# 1.current class 2.parent class (left to right) 3.Grand parents 4.object(class)
```

```python
# define MRO ---> the order in which python searches for methods in a class hierarchy
# which algorithm is used? C3 Linearization
# why MRO is important? to resolve ambiguity in multiple inheritance  
# Tips:  --->left to right ---> no repetition  ---> parent before child
```

```python
# define C3 Linearization --> It is the algorithm in python uses to compute MRO to compute in multiple inheritance 
# ---> It creates one consistent order in search for methods 
```

```python
#why? --> avoids ambiguity --> solves the diamond problem ---> ensures parent before child rule --> supports super()
# key rules --> child before parent,left to right order,no class appears twice ,parent order is preserved 
```

```python
# C3 Linearization Formula :
# For class C(A,B):
# MRO[C]=[C]+merge(MRO(A),MRO(B),[A,B])
```

```python
class A:
    pass
class B(A):
    pass
class C(A):
    pass
class D(B,C):   
    pass
# If we can create this it shows type error and it cannot create consistent method resoultion order(mro) and parent order conflict violates C3 rules.
```

```python
# Step-1 write the mro of box classes
# MRO(A)=[A,object]
# MRO(B)=[B,A,object]
# MRO(C)=[C,A,object]
#step-2 Apply c3 formula
# MRO(D)=[D]+merge([B,A,object],[C,A,object],[B,C])
# step-3 Merge process
# 1.pick B(not in tail of any list) 2.pick C 3.pick A 4.pick object
```

```python
# Hierarchical Inheritance
# From one parent to multiple children to acquire the properties
class parent:
    pass
class child1(parent):
    pass
class child2(parent):
    pass
```

```python
# Hybrid Inheritance
# combination of more than one type of inheritance.
```

```python
# super()--> It is used to call methods or constructors of the parent class from the child class and it  is mostly used to call parent constructor
# syntax:super().__init__(arguments)
```

```python
# without super
class Parent:
    def __init__(self):
        print("Parent Constructor")
class Child(Parent):
    def __init__(self):
        print("Child Constructor")
Child()
```

```python
# with super
class Parent:
    def __init__(self):
        print("Parent Constructor")
class Child(Parent):
    def __init__(self):
        super().__init__()
        print("Child Constructor")
Child()
```

```python
# Parent constructor with parameters
class Parent:
    def __init__(self,name):
        self.name=name
class Child(Parent):
    def __init__(self,name,age):
        super().__init__(name)
        self.age=age
c=Child("Ganesh",23)
c.name,c.age
```

```python
#calling parent method using super()
class Parent:
    def show(self):
        print("Parent Method")
class Child(Parent):
    def show(self):
        super().show()  # call parent method
        print("Child Method")
Child().show()
```

```python
# super() --> Multiple Inheritance ---> MRO Order
class A:
    def __init__(self):
        print('A')
class B(A):
    def __init__(self):
        super().__init__()
        print('B')
class C(B):
    def __init__(self):
        super().__init__()
        print('C')
C()
```

```python
# Method Overriding: 
# It occurs when a child class provides its own implementation of a method
# that already exists in the parent class using the same method name and parameters 
# It is the type of runtime polymorphism 
```

```python
# why? to change the parent class behavior,to provide specialize implementation
# method name  and parameters must be same,Inheritance must exist ,happens at runtime
```

```python
class Parent:
    def show(self):
        print("Parent Method")
class Child(Parent):
    def show(self): # overriding
        print("Child Method")
c=Child()
c.show()     
```

```python
# Calling parent method using super()
class Parent:
    def show(self):
        print("Parent Method")
class Child(Parent):
    def show(self):
        super().show()
        print("Child Method")
c=Child()
c.show()
```

```python
# Overriding with constructor
class Parent:
    def __init__(self):
        print("Parent Method")
class Child(Parent):
    def __init__(self):
        super().__init__()
        print("Child Method")
Child()
```

### Problem: Child class __init__ doesn’t call parent __init__, so parent attributes aren’t initialized.

```python
class Parent:
    def __init__(self):
        self.x = 10
 
class Child(Parent):
    def __init__(self):
        # ❌ Missing super().__init__()!
        pass
 
c = Child()
print(c.x)  # AttributeError: 'Child' object has no attribute 'x'

# super() method and constructor is missing-This is really important to intialize the paren attributes in the child class.
```

### Problem: Two parent classes inherit from the same grandparent → ambiguity.

```python
class A:

    def method(self):

        print("A")
 
class B(A):

    def method(self):

        print("B")
 
class C(A):

    def method(self):

        print("C")
 
class D(B, C):# whichever methods calls first then it is the output

    pass
 
d = D()

d.method()  # Calls B.method() due to MRO order

 
```

```python
print(D.mro())# order of the execution is the same if you run myltiple times
```

```python
class Parent:
    def method(self, x, y):
        print(f"Parent: {x + y}")
 
class Child(Parent):
    def method(self, x):  # This contributes to method overloading based on the method signature.(python doesn't have method overloading it is just for understanding)
        super().method(x, x)
c1=Child()
c1.method(3)
# TypeError: method() missing 1 required positional argument: 'y'
```

### Problem: Child class doesn’t call parent’s string representation.

```python
class Parent:
    def __repr__(self):
        return "Parent()"
 
class Child(Parent):
    def __repr__(self):
        return f"Child({super().__repr__()})"  # ❌ No parent info
 
c = Child()
print(c)  # Child() ❌ Missing parent details
```

```python
print(Child.mro())
```

```python
class Parent:

    def __init__(self, data=[]):  # ❌ Shared mutable default!

        self.data = data
 
class Child(Parent):

    pass
 
c1 = Child()

c1.data.append(1)

c2 = Child()

print(c2.data)  # [1] ❌ Unexpected!

 
```
