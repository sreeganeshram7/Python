```python
class EncapsulationExample:
    def __init__(self):
        # Private attribute
        self.__private_var = 10
        # Protected attribute
        self._protected_var = 20

    # Getter for private attribute
    def get_private_var(self):
        return self.__private_var

    # Setter for private attribute
    def set_private_var(self, value):
        self.__private_var = value

    # Getter for protected attribute
    def get_protected_var(self):
        return self._protected_var

    # Setter for protected attribute
    def set_protected_var(self, value):
        self._protected_var = value
```

```python

# Creating an instance of the class
obj = EncapsulationExample()

# Accessing private and protected attributes using public methods
print(obj.get_private_var())      # Output: 10
print(obj.get_protected_var())    # Output: 20

# Modifying private and protected attributes using public methods
obj.set_private_var(30)
obj.set_protected_var(40)

# Displaying the modified values
print(obj.get_private_var())      # Output: 30
print(obj.get_protected_var())    # Output: 40
```

```python
class Human:
    def __init__(self, age):
        # This calls the 'age' setter method below, not the internal _age attribute directly
        self._age = age
 
    ## @property
    def get_age(self):
        """The age property: allows getting the internal age."""
        print("Getting value...")
        return self._age # Accesses the internal private attribute
 
    ## @age.setter
    def set_age(self, value):
        """Allows setting the internal age with validation."""
        print(f"Setting value to {value}...")
        if not isinstance(value, int):
            raise TypeError("Age must be an integer.")
        if value < 0:
            raise ValueError("Age cannot be negative.")
        self._age = value
 
    ## @age.deleter
    def age(self):
        """Allows deleting the age attribute."""
        print("Deleting value...")
        del self._age
```

```python
person = Human(26) 
print(person._age) 
print(person.get_age())
person.set_age(32)
print(person.get_age())
```

## Challenge: Create a Temperature class that stores a private _celsius value.
### Implement a celsius property.
### Add a setter that ensures the temperature cannot be set below absolute zero (-273.15°C). If it is, raise a ValueError.
#### Goal: Practice using setters to protect internal state from invalid data. 

```python
class Temperature:
    def __init__(self,celsius):
        self.__celsius=celsius
    @property
    def celsius(self):
        print("Getting value ...")
        return self.__celsius
    @celsius.setter
    def celsius(self,value):
        print(f"Setting value to {value}...")
        if value<-273.15:
            raise ValueError("Don't set the temperature below absoulte zero")
        self.__celsius=value
```

```python
temp=Temperature(25)
print(temp.celsius)
```

```python
temp.celsius=105
try:
    temp.celsius=-300
except(ValueError) as e:
    print(e)
```

## Challenge: Create a User class with private _first_name and _last_name. 
### Create a fullname property that returns the combined names (e.g., "John Doe").
### Add a setter for fullname that takes a single string (e.g., "Jane Smith"), splits it, and updates the individual _first_name and _last_name attributes.
### Goal: Practice managing multiple internal attributes through a single external property.

```python
class User:
    def __init__(self,first_name,last_name):
        self.__first_name=first_name
        self.__last_name=last_name
    @property
    def fullname(self):
        print("Getting Value ...")
        return f"{self.__first_name} {self.__last_name}"
    @fullname.setter
    def fullname(self,value):
        if isinstance(value, tuple):
            raise TypeError("Pass full name as a single string, not tuple")
        names=value.split()
        if len(names)!=2:
            raise ValueError("Full name must contains exactly two words")
        self.__first_name=names[0]
        self.__last_name=names[1]
```

```python
u=User("ganesh","ram")
print(u.fullname)
```

```python
u.fullname="vishal balaji"
try:
    u.fullname="shree","vishal","balaji"
except(TypeError,ValueError) as e:
    print(e)
```
