```python
#  Rotate List to Right by K Steps
nums=list(map(int,input().split()))
k=int(input())
n=len(nums)
k=k%n
def reverse(arr,start,end):
    while start<end:
        arr[start],arr[end]=arr[end],arr[start]
        start+=1
        end-=1
reverse(nums,0,n-1)
reverse(nums,0,k-1)
reverse(nums,k,n-1)
print(nums)
```

```python
# Product of Array Except Self (LeetCode Medium)
# Given a list nums, return a new list answer such that:answer[i] = product of all elements except nums[i]
nums = list(map(int, input().split()))
n=len(nums)
ans=[1]*n
left=1
for i in range(n):
    ans[i]=left
    left*=nums[i]
right=1
for i in range(n-1,-1,-1):
    ans[i]*=right
    right*=nums[i]
print(ans)
```

```python
# Trapping rain water
height=list(map(int,input().split()))
start=0
end=len(height)-1
l=0
r=0
total=0
while start<end:
    l=max(l,height[start])
    r=max(r,height[end])
    if l<=r:
        total+=l-height[start]
        start+=1
    else:
        total+=r-height[end]
        end-=1
print(total)
```

```python
# Factorial of a Number
def factorial(n):
    fact=1
    for i in range(1,n+1):
        fact*=i
    return fact
```

```python
# Prime Number
def PrimeNumber(n):
    for i in range(2,n):
        if n%i==0:
            return "Not Prime"
            break
    else:
        return "Prime"
```

```python
# Count the digits
def countdigits(n):
    c=0
    while n>0:
        c+=1
        n//=10
    return c
        
```

```python
# First Non-Repeating Character
def NonRepeating(s):
    for i in s:
        if s.count(i)==1:
            return i
    return None  
```

```python
# GCD of two numbers
def gcd(a,b):
    for i in range(1,min(a,b)+1):
        if a%i==0 and b%i==0:
            gcd=i
    return gcd
# HCF of two numbers
def lcm(a,b):
    return (a*b)//gcd(a,b)
    
```

```python
gcd(4,8)
```

```python
lcm(4,8)
```

```python
# Longest substring without repeating the characters
s=input()
seen=set()
left=0
max_len=0
for right in range(len(s)):
    if s[right] in seen:
        seen.remove(s[left])
        left+=1
    seen.add(s[right])
    max_len=max(max_len,right-left+1)
print(max_len)
    
```

```python
# Maximum Average Subarray I(lc-643)
nums=list(map(int,input().split()))
k=int(input())
window_sum=sum(nums[:k])
max_sum=window_sum
for i in range(k,len(nums)):
    window_sum=window_sum-nums[i-k]+nums[i]
    max_sum=max(max_sum,window_sum)
print(max_sum/k)
```

```python
# Minimum Size Subarray Sum(lc-209)
nums=list(map(int,input().split()))
target=int(input())
left=0
total=0
min_len=float('inf')
for right in range(len(nums)):
    total+=nums[right]
    while total>=target:
        min_len=min(min_len,right-left+1)
        total-=nums[left]
        left+=1
if min_len==float('inf'):
    print(0)
else:
    print(min_len)
```
