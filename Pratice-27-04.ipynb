```python
# Longest substring without repeating the characters
s=input()
c={}
left=0
max_len=0
for right in range(len(s)):
    if s[right] in c:
        left=max(left,c[s[right]]+1)
    c[s[right]]=right
    max_len=max(max_len,right-left+1)
print(max_len)
```

```python
# Valid anagram
s=input()
t=input()
if len(s)!=len(t):
    print(False)
c={}
for i in s:
    c[i]=c.get(i,0)+1
for i in t:
    if i not in c or c[i]==0:
        print(False)
        break
else:
    print(True)
```

```python
# first non repeating character
s=input()
freq={}
for i in s:
    freq[i]=freq.get(i,0)+1
for i in s:
    if freq[i]==1:
        print(i)
        break
```

```python
# longest repeating characters replacement
s=input()
k=int(input())
l=0
mf=0
ml=0
freq={}
for r in range(len(s)):
    freq[s[r]]=freq.get(s[r],0)+1
    mf=max(mf,freq[s[r]])
    if (r-l+1)-mf>k:
        freq[s[l]]-=1
        l+=1
    ml=max(ml,r-l+1)
print(ml)
```

```python
# Group Anagrams
s=input().split()
d={}
for i in s:
    key="".join(sorted(s))
    if key not in d:
        d[key]=[]
    d[key].append(i)
print(list(d.values()))   
```

```python
### Two sum
nums=list(map(int,input().split()))
t=int(input())
seen={}
for i,num in enumerate(nums):
    if t-num in seen:
        print(seen[t-num],i)
    seen[num]=i
```

```python
# Best time to buy and sell stock
nums=list(map(int,input().split()))
a=float('inf')
b=0
for i in nums:
    a=min(a,i)
    b=max(b,i-a)
print(b)
```

```python
# product of array except self
nums=list(map(int,input().split()))
n=len(nums)
ans=[1]*n
l=1
for i in range(n):
    ans[i]=l
    l*=nums[i]
r=1
for i in range(n-1,-1,-1):
    ans[i]*=r
    r*=nums[i]
print(ans)
```

```python
##### Maximum subarray
nums=list(map(int,input().split()))
c=nums[0]
m=nums[0]
for i in range(1,len(nums)):
    c=max(nums[i],c+nums[i])
    m=max(m,c)
print(m)
```

```python
# minimum size subarray sum
nums=list(map(int,input().split()))
t=int(input())
left=0
total=0
min_len=float('inf')
for right in range(len(nums)):
    total+=nums[right]
    while total>=t:
        min_len=min(min_len,right-left+1)
        total-=nums[left]
        left+=1
if min_len==float('inf'):
    print(0)
else:
    print(min_len)
```

```python
# Subarray sum equals k
arr=list(map(int,input().split()))
k=int(input())
window_sum=sum(arr[:k])
max_sum=window_sum
for i in range(k,len(arr)):
    window_sum+=arr[i]-arr[i-k]
    max_sum=max(max_sum,window_sum)
print(max_sum)
```
