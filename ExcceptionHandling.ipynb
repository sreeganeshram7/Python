```python
#syntax -> 1.try:#risky code 2.except:ExceptionType:#handle error 3.else:#runs if no exception 4.finally:always execute
```

```python
try:
    x=int(input("enter number:"))
    print(10/x)
except ZeroDivisonError:
    print("cannot divide by zero")
except ValueError:
    print("Invalid input")
else:
    print("success")
finally:
    print("Program Finished")
```

```python
# custom exception - it is a user defined exception created to handle specific application errors.it is created by inheriting from exception class
```

```python
#why? 1.meaningful error messages 2.better debugging 3.application-specific validation 4.cleaner code structure
```
