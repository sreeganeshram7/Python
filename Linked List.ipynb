Linked List:“A Linked List is a linear data structure where elements are stored in nodes, and each node contains data and a reference (pointer) to the next node.
Unlike arrays, elements are not stored in contiguous memory.”
Structure:
Each node has:data,next
The list has:Head → starting node,Last node → points to None
Types:
Singly Linked List → one direction
Doubly Linked List → next + previous
Circular Linked List → last connects to first
Linked List vs Array (Key Interview Point)
Arrays → fast access (O(1))
Linked List → fast insertion/deletion (O(1) at head)
Linked List → dynamic size
Arrays → fixed / resizing costly
When Do We Use It?
When frequent insertions/deletions are needed
When memory flexibility is required
Used in stacks, queues, graphs
drawback:
One drawback is that accessing elements is slow because we must traverse from the head, which takes O(n) time.”

A linked list is a linear data structure made of nodes, where each node stores data and a pointer to the next node. Unlike arrays, it doesn’t store elements in contiguous memory, which makes insertion and deletion efficient. However, accessing elements is slower because traversal is required. There are types like singly, doubly, and circular linked lists.”

```python
class LinkedList:
    def __init__(self):
        self.head = None

    # Insert at beginning
    def insert_begin(self, data):
        new_node = Node(data)
        new_node.next = self.head
        self.head = new_node

    # Insert at end
    def insert_end(self, data):
        new_node = Node(data)
        if not self.head:
            self.head = new_node
            return
        
        temp = self.head
        while temp.next:
            temp = temp.next
        
        temp.next = new_node

    # Print list
    def display(self):
        temp = self.head
        while temp:
            print(temp.data, end=" -> ")
            temp = temp.next
        print("None")
```

```python
ll = LinkedList()

ll.insert_begin(10)
ll.insert_begin(5)
ll.insert_end(20)

ll.display()
```

```python
# Traversal and logic: Search Element
class LinkedList:
    def __init__(self):
        self.head=None
        
    def insert_end(self, data):
        new_node = Node(data)
        if not self.head:
            self.head = new_node
            return
        
        temp = self.head
        while temp.next:
            temp = temp.next
        
        temp.next = new_node 
        
    def search(self,key):
        temp=self.head
        while temp:
            if temp.data==key:
                return True
            temp=temp.next
        return False
```

```python
ll = LinkedList()

ll.insert_end(5)
ll.insert_end(10)
ll.insert_end(20)
```

```python
# Find Maximum Element
class LinkedList:
    def __init__(self):
        self.head=None
        
    def insert_end(self, data):
        new_node = Node(data)
        if not self.head:
            self.head = new_node
            return
        
        temp = self.head
        while temp.next:
            temp = temp.next
        
        temp.next = new_node 
        
    def find_max(self):
        if not self.head:
            return None
        max_val=self.head.data
        temp=self.head
        while temp:
            if temp.data>max_val:
                max_val=temp.data
            temp=temp.next
        return max_val
        
    # delete first Node
    def delete_first(self):
        if self.head:
            self.head=self.head.next
            
    # delete last Node
    def delete_last(self):
        if not self.head:
            return
        if not self.head.next:
            self.head=None
            return
        temp=self.head
        while temp.next.next:
            temp=temp.next
        temp.next=None
    # Print list
    def display(self):
        temp = self.head
        while temp:
            print(temp.data, end=" -> ")
            temp = temp.next
    print("None")
```

```python
ll = LinkedList()
ll.insert_end(5)
ll.insert_end(10)
ll.insert_end(20)
print(ll.display())
print(ll.find_max())
```
