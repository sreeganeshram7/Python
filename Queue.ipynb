“Queue is a FIFO data structure where elements are inserted at the rear and removed from the front. It is commonly used in real-world systems like printer queues, CPU scheduling, and BFS traversal.”

why deque? “Because list pop(0) is O(n), but deque provides O(1) insertion and deletion from both ends.”

```python
# 1. Print Queue Elements 1 2 3 4 5
from collections import deque
nums=list(map(int,input().split()))
q=deque(nums)
for i in q:
    print(i,end=" ")
```

```python
# 2.Remove All Elements (Deque)
from collections import deque 
q=deque(map(int,input().split()))
while q:
    print(q.popleft(),end=" ")
```

```python
# 3. Print Front Element
from collections import deque 
q=deque(map(int,input().split()))
print(q[0])
```

```python
# 4. Check Queue Empty or Not
from collections import deque
q=deque()
if not q:
    print("Empty")
```

```python
# 5. Reverse a Queue
from collections import deque
q = deque(map(int, input().split()))
stack = []
# Step 1: Push into stack
while q:
    stack.append(q.popleft())
# Step 2: Push back into queue
while stack:
    q.append(stack.pop())

print("Reversed Queue:", list(q))
```

approach for this qn:
Enqueue(x)
Push into stack1
Dequeue()
If stack2 empty:
Move all elements from stack1 → stack2
Pop from stack2

```python
# Implement queue using stack
class MyQueue:
    def __init__(self):
        self.stack1 = []
        self.stack2 = []

    def enqueue(self, x):
        self.stack1.append(x) 

    def dequeue(self):
        if not self.stack2:
            while self.stack1:
                self.stack2.append(self.stack1.pop())
        return self.stack2.pop()

    def peek(self):
        if not self.stack2:
            while self.stack1:
                self.stack2.append(self.stack1.pop())
        return self.stack2[-1]

    def isEmpty(self):
        return not self.stack1 and not self.stack2
```

```python
q = MyQueue()

q.enqueue(10)
q.enqueue(20)
q.enqueue(30)

print(q.dequeue())  
print(q.peek())  
```
