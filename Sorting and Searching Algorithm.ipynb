Bubble Sort

```python
arr=list(map(int,input().split()))
for i in range(0,len(arr)-1):
    for j in range(0,len(arr)-1):
        if arr[j]<arr[j+1]:
            continue
        elif arr[j]>arr[j+1]:
            arr[j],arr[j+1]=arr[j+1],arr[j]
print(arr)
```

```python
arr=list(map(int,input().split()))
while True:
    a=True
    for j in range(0,len(arr)-1):
        if arr[j]>arr[j+1]:
            arr[j],arr[j+1]=arr[j+1],arr[j]
            a=False
    if a==True:
        break
print(arr)
```

```python
arr = list(map(int, input().split()))
n = len(arr)

for i in range(n - 1):
    swapped = False
    
    for j in range(0, n - i - 1):   # reduced range
        if arr[j] > arr[j + 1]:
            arr[j], arr[j + 1] = arr[j + 1], arr[j]
            swapped = True
    
    if not swapped:   # already sorted
        break

print(arr)
```

Selection sort

```python
arr=list(map(int,input().split()))
for pos in range(len(arr)-1):
    min=pos
    for i in range(pos+1,len(arr)):
        if arr[i]<arr[min]:
            min=i
    arr[min],arr[pos]=arr[pos],arr[min]
print(arr)
```

Insertion Sort

```python
arr=list(map(int,input().split()))
for i in range(1,len(arr)):
    cur=arr[i]
    j=i-1
    while j>=0 and cur<arr[j]:
        arr[j+1]=arr[j]
        j-=1
    arr[j+1]=cur
print(arr)
```

Linear Search

```python
arr=list(map(int,input().split()))
x=int(input())
for i in range(len(arr)):
    if arr[i]==x:
        print(i)
        break
else:
    print("x is not present")
```

Binary Search

```python
arr=list(map(int,input().split()))
x=int(input())
l=0
r=len(arr)-1
while l<=r:
    m=(l+r)//2
    if x==arr[m]:
        print(m)
        break
    elif x>arr[m]:
        l=m+1
    elif x<arr[m]:
        r=m-1
else:
    print(-1)
```
