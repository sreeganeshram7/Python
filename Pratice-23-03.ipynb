Mirror number

```python
num=int(input())
temp=num
rev=0
while num>0:
    r=num%10
    rev=rev*10+r
    num=num//10
if temp==rev:
    print("Mirror Number")
else:
    print("Not a Mirror Number")
```

```python
n = int(input("Enter number: "))
temp = n
sum_fact = 0

# function to calculate factorial
def fact(x):
    f = 1
    for i in range(1, x+1):
        f *= i
    return f

while n > 0:
    digit = n % 10
    sum_fact += fact(digit)
    n = n // 10

if sum_fact == temp:
    print("Strong Number")
else:
    print("Not a Strong Number")
    
```

```python
n = int(input("Enter number: "))

if n < 0:
    print("Enter a positive number")
else:
    temp = n
    sum_fact = 0

    while n > 0:
        digit = n % 10
        
        # factorial using loop
        fact = 1
        for i in range(1, digit + 1):
            fact *= i
        
        sum_fact += fact
        n = n // 10

    if sum_fact == temp:
        print("Strong Number")
    else:
        print("Not a Strong Number")
```

```python
n = int(input("Enter number: "))

freq = {}

while n > 0:
    digit = n % 10
    
    if digit in freq:
        freq[digit] += 1
    else:
        freq[digit] = 1
    
    n = n // 10

print("Digit Frequency:")
for key in sorted(freq):
    print(key, "→", freq[key])
```

```python
n = int(input("Enter number: "))


if n == 0:
    print("0 → 1")
else:
    freq = [0] * 10   
    
    while n > 0:
        digit = n % 10
        freq[digit] += 1
        n = n // 10

    print("Digit Frequency:")
    for i in range(10):
        if freq[i] > 0:
            print(i, "→", freq[i])
```

remove duplicates

```python
n=int(input())
res=0
place=1
while n>0:
    digit=n%10
    t=res
    found=False
    while t>0:
        if t%10==digit:
            found=True
        t=t//10
    if not found:
        res=res+digit*place
        place*=10
    n=n//10
print(res)
```

```python

n = int(input("Enter number: "))

result = 0
place = 1

while n > 0:
    digit = n % 10
    
    # check if digit already exists in result
    temp = result
    found = False
    
    while temp > 0:
        if temp % 10 == digit:
            found = True
            break
        temp = temp // 10
    
    # if not found, add to result
    if not found:
        result = result + digit * place
        place *= 10
    
    n = n // 10

print("Without duplicates:", result)
```

```python
n = int(input())

if n < 0:
    print("Enter positive number")
else:
    maximum = -1
    second = -1

    while n > 0:
        res = n % 10

        if res > maximum:
            second = maximum
            maximum = res
        elif res > second and res != maximum:
            second = res

        n //= 10

    if second == -1:
        print("No second largest")
    else:
        print(second)
```

```python
s = input()

i = 0
j = len(s) - 1

is_palindrome = True

while i < j:
    if s[i] != s[j]:
        is_palindrome = False
        break
    i += 1
    j -= 1

if is_palindrome:
    print("Palindrome")
else:
    print("Not Palindrome")
```
