```python
# properties ----> it is used to access instance variables using methods it looking like normal attributes
# ----> It is used to acheive encapsulation(data hiding)
```

```python
# @property or @property getter----> it allows you to access methods like an attribute
# @property_name_setter ---> It is used to define a setter method for a property created using @property.
                           # It allows you to control how a value is assigned to an attribute.
# @property_name.deleter ---> it is used to define what happens when a property is deleted . 
                         # It gives controlled deletion of an attribute.
# Decorator is a function adds extra behavior to another function or method without changing its code
# attributes ---> it stores data hierarchy
```

```python
# getter ---> It is a method used to read/access the value of a private attribute dnd it returns the value
# setter ---> It is a method used to modify/update the value of a private attribute and it sets/updates the value
# deleter ---> It is a decorator used to define the behavior that occurs when property name is deleted using del obj.name 
```

```python
# property(getter+setter)
class Employee:
    def __init__(self,salary):
        self.__salary=salary  # private attribute
    @property
    def salary(self):
        return self.__salary # getter
    @salary.setter # setter 
    def salary(self,value):
        if value <0:
            print("Salary cannot be negative")
        else:
            self.__salary=value
    @salary.deleter #deleter
    def salary(self):
        print("Salary Deleted")
        del self.__salary
e=Employee(50000)
e.salary         # getter
```

```python
e.salary=60000   # setter 
e.salary
```

```python
del e.salary   # deleter
```

```python
class Student:
    def __init__(self,age):
        self.__age=age
    @property
    def age(self):
        return self.__age
    @age.setter
    def age(self,value):
        if value<=0:
            print("Please give me age value in positive")
        else:
            self.__age=value
    @age.deleter
    def age(self):
        print("age deleted")
        del self.__age
```
