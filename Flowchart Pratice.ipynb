## FCP047-HCF of N Numbers
#### Write a program to take N numbers from the user and print the Highest Common Factor (HCF) of the given set of numbers. Take N from the user as input.

```python
n = int(input())

nums = []
for i in range(n):
    nums.append(int(input()))

def hcf(a, b):
    while b != 0:
        a, b = b, a % b
    return a

result = nums[0]

for num in nums[1:]:
    result = hcf(result, num)

print(result)
```

Use the following algorithms to sort the array - 
Bubble Sort   

```python
import array
n=int(input())
arr=array.array('i',[])
for _ in range(n):
    arr.append(int(input()))
for i in range(n-1):
    for j in range(n-1):
        if arr[j]>arr[j+1]:
            arr[j],arr[j+1]=arr[j+1],arr[j]
for i in arr:
    print(i)
```

FCP070	Binary Search on Sorted Array	"Write a program to take a sorted array from the user as input and find a number using the Binary Search algorithm."

```python
import array
n=int(input())
arr=array.array('i',[])
for _ in range(n):
    arr.append(int(input()))
x=int(input())
l=0
r=len(arr)-1
found=False
while l<=r:
    m=(l+r)//2
    if x==arr[m]:
        found=True
        break
    elif x>arr[m]:
        l=m+1
    elif x<arr[m]:
        r=m-1
if found==True:
    print("Yes")
else:
    print("No")
```

FCP095	Insertion sort	Take N numbers from the user as input and print them in an ascending order. 
Use the following algorithms to sort the array - 
Insertion Sort

```python
import array
n=int(input())
arr=array.array('i',[])
for _ in range(n):
    arr.append(int(input()))
for i in range(1,n):
    cur=arr[i]
    j=i-1
    while j>=0 and cur<arr[j]:
        arr[j+1]=arr[j]
        j-=1
    arr[j+1]=cur
for i in arr:
    print(i)   
```

FCP096	Selection Sort	"Take N numbers from the user as input and print them in an ascending order. 
Use the following algorithms to sort the array - 
Selection  Sort

```python
import array
n=int(input())
arr=array.array('i',[])
for _ in range(n):
    arr.append(int(input()))
for pos in range(len(arr)-1):
    min=pos
    for i in range(pos+1,len(arr)):
        if arr[i]<arr[min]:
            min=i
    arr[min],arr[pos]=arr[pos],arr[min]
for i in arr:
    print(i)
```
