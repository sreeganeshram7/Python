Elements are inserted and removed from the top using push and pop operations.
It is commonly used in problems like parentheses validation, recursion, and undo operations.”

Problem: Valid Parentheses:Given a string:"()[]{}": Check if it is valid

```python
def is_valid(s):
    stack = []
    
    for ch in s:
        if ch in "({[":
            stack.append(ch)
        else:
            if not stack:
                return False
            
            top = stack.pop()
            
            if (ch == ')' and top != '(') or \
               (ch == '}' and top != '{') or \
               (ch == ']' and top != '['):
                return False
    
    return len(stack) == 0


# Test
print(is_valid("()[]{}"))   # True
print(is_valid("(]"))   # False
print(is_valid("({[]})")) # True
```

```python
def is_valid(s):
    stack = []
    pairs = {')': '(', '}': '{', ']': '['}

    for ch in s:
        if ch in "({[":
            stack.append(ch)
        else:
            if not stack or stack.pop() != pairs[ch]:
                return False

    return len(stack) == 0
# Test
print(is_valid("()[]{}"))   # True
print(is_valid("(]"))   # False
print(is_valid("({[]})")) # True
```

```python
print(is_valid("((()))"))
print(is_valid("([{}])"))
print(is_valid("((())"))
print(is_valid("{[()]}"))
```

```python
def is_balanced(s):
    stack = []

    for ch in s:
        if ch=="(":
            stack.append(ch)
        else:
            if not stack:
                return False
            stack.pop()

    return len(stack) == 0
```

```python
# Test
print(is_balanced("(()())"))  # True
print(is_balanced("((())"))   # False
print(is_balanced("())("))    # False
```

A Min Stack is a stack that supports all normal operations PLUS:
Get minimum element in O(1) time
push(x) → push element
pop() → remove top
top() → get top element
getMin() → return minimum element
Normal stack gives top in O(1)
 But finding minimum normally = O(n) ❌

So we use:
Two stacks

Main stack
Min stack (to track minimums)

```python
# Min Stack
class MinStack:
    def __init__(self):
        self.stack = []
        self.min_stack = []

    def push(self, val):
        self.stack.append(val)
        
        if not self.min_stack or val <= self.min_stack[-1]:
            self.min_stack.append(val)

    def pop(self):
        if self.stack[-1] == self.min_stack[-1]:
            self.min_stack.pop()
        self.stack.pop()

    def top(self):
        return self.stack[-1]

    def getMin(self):
        return self.min_stack[-1]
```

```python
s = MinStack()
s.push(5)
s.push(3)
s.push(7)
s.push(2)

print(s.getMin())  # 2
s.pop()
print(s.getMin())  # 3
```

```python
# Remove Adjacent Duplicates
s=input()
stack=[]
for i in s:
    if stack and stack[-1]==i:
        stack.pop()
    else:
        stack.append(i)
r="".join(stack)
print(r)
```

```python
# Important Pattern:Next Greater Element
# Problem:Given an array, for each element find the next greater element to the right
arr=list(map(int,input().split()))
s=[]
r=[-1]*len(arr)
for i in range(len(arr)-1,-1,-1):
    while s and s[-1]<=arr[i]:
        s.pop()
    if s:
        r[i]=s[-1]
    s.append(arr[i])
print(r)
```

```python
# Problem: Next Greater Element (LEFT):For each element, find the first greater element on the LEFT side
arr=list(map(int,input().split()))
s=[]
r=[-1]*len(arr)
for i in range(len(arr)):
    while s and s[-1]<=arr[i]:
        s.pop()
    if s:
        r[i]=s[-1]
    s.append(arr[i])
print(r)
```
