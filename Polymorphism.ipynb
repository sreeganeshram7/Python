```python
# Define Polymorphism-It means one name but many forms
# the same method or operation behaves differently depending on the object
# It enables the flexibility and dynamic behavior in programs
```

```python
# why?
# It reduces codes duplication ,It improves flexibility,It supports scalability,It enables dynamic method calls,It makes code more maintainable
```

```python
#Types
# 1.Compile time polymorphism(Method overloading-Limited in python)
# ---> It doesn't supports the true method of overloading but it can be acheived by using default arguments
class Math:
    def add(self,a,b=0,c=0):
        return a+b+c
m=Math()
print(m.add(2,3))
print(m.add(2,3,4))
```

```python
# 2.Run time polymorphism(method overriding)
# child class overrides parent class method
class Animal:
    def speak(self):
        print("Animal Sound")
class Dog(Animal):
    def speak(self):
        print("Dog barks")
class Cat(Animal):
    def speak(self):
        print("Cat meows")
animals=[Dog(),Cat()]
for a in animals:
    a.speak()
```

```python
# Built in Polymorphism: same function but different behavior
print(len("Python"))
print(len([1,2,3]))
```

```python
# Operator Overloading: it works differently based on type
print(10+20)
print("Hello" + "World")
```

```python
#Duck Typing:Python specific polymorphism-if it behaves like a duck
# Duck Typing is a concept in Python where the type of an object is not checked. 
# Instead, Python checks whether the object has the required methods or behavior.
class Bird:
    def fly(self):
        print("Bird flies")
class Plane:
    def fly(self):
        print("Plane flies")
def make_fly(obj):
    obj.fly()
    
make_fly(Bird())
make_fly(Plane())
```
