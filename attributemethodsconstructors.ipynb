```python
# Class Attribute ---> It is shared by all objects because it is defined in the outside of the class
class Student:
    school='Boys Town'
```

```python
# Instance attribute ---> It is belongs to the object
class Student:
    def __init__(self,name,age):
        self.name=name
        self.age=age
```

```python
def display(self):# Instance method --> It is working with object data
    print(self.name,self.age)
```

```python
@classmethod # It is working with class data
def get_school(cls):
    return cls.school
```

```python
@staticmethod # it is used for utility logic
def is_pass(marks):
    return marks>=35
```

```python
class Student:
    school='BoysTown'
    def __init__(self,name,age):
        self.name=name
        self.age=age
    def display(self):
        print(self.name,self.age)
    @classmethod
    def school_name(cls):
        return cls.school
    @staticmethod
    def is_pass(marks):
        return marks>=35
s1=Student('Bala',20)
s1.display()
print(s1.school_name())
s1.is_pass(80)
```

```python
# Default Constructor   -- It takes no parameters except self
class Student:
    def __init__(self):
        self.name='Unknown'
        self.age=0
s1=Student()
print(s1.name,s1.age)
```

```python
# Parameterized constructor  -- It takes parameters
class Sports:
    def __init__(self,name,team):
        self.name=name
        self.team=team
s1=Sports('MSD','CSK')
print(s1.name,s1.team)
```

```python
# constructor with default arguments  ---It includes default values
class Shop:
    def __init__(self,name='NA',brand='NA'):
        self.name=name
        self.brand=brand
s1=Shop()
s2=Shop('Panipuri','SS')
s1.name
s2.brand
```

```python
# Copy Constructor  -- It crates new object
class Shop:
    def __init__(self,name,brand):
        self.name=name
        self.brand=brand
    @classmethod
    def copy(cls,other):
        return cls(other.name,other.brand)
s1=Shop('Panipuri','SS')
s2=Shop.copy(s1)
s2.name
```
