```python
# swapping method
n = int(input("Enter size: "))
arr = []

for i in range(n):
    arr.append(int(input()))

pos = int(input("Enter position: "))
val = int(input("Enter value: "))

prev = val
i = pos

while i <= n:
    if i < len(arr):
        temp = arr[i]
        arr[i] = prev
        prev = temp
    else:
        arr.append(prev)
    i += 1

print(arr)
```

```python
# shifting method
n = int(input("Enter size: "))
arr = []

for i in range(n):
    arr.append(int(input()))

pos = int(input("Enter position: "))
val = int(input("Enter value: "))

arr += [0]
i = n
while i > pos:
    arr[i] = arr[i-1]
    i -= 1

arr[pos] = val

print(arr)
```

```python
# deletion
n = int(input("Enter size: "))
arr = []
for i in range(n):
    arr.append(int(input()))
pos = int(input("Enter position to delete: "))
i = 0
k = 0
while i < n:
    if i != pos:
        arr[k] = arr[i]
        k += 1
    i += 1
n = n - 1
i = 0
while i < n:
    print(arr[i], end=" ")
    i += 1
```

```python
# MIN AND MAX
n = int(input("Enter size: "))
arr = []

for i in range(n):
    arr.append(int(input()))

min_val = arr[0]
max_val = arr[0]

i = 1
while i < n:
    if arr[i] > max_val:
        max_val = arr[i]
    if arr[i] < min_val:
        min_val = arr[i]
    i += 1

print("Max =", max_val)
print("Min =", min_val)
```

```python
# Frequency
n = int(input("Enter size: "))
arr = []

for i in range(n):
    arr.append(int(input()))

i = 0
while i < n:
    if arr[i] != -1:
        count = 0
        val = arr[i]

        j = i
        while j < n:
            if arr[j] == val:
                count += 1
                arr[j] = -1 
            j += 1

        print(val, "=", count)

    i += 1
```

```python
# pythagorean tripelts
n = int(input())

a = 1
while a <= n:
    b = a
    while b <= n:
        c = b
        while c <= n:
            if a*a + b*b == c*c:
                print(a, b, c)
            c += 1
        b += 1
    a += 1
```
