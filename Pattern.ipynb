## Shrinking-1

```python
n=int(input())
for i in range(n,0,-1):
    print("* "*i)
```

```python
n=int(input())
for i in range(n,0,-1):
    for j in range(i):
        print("*",end="")
    print()
```

## Number Pattern

```python
# Same Star:
n=int(input())
for i in range(1,n+1): # for i in range(n)
    print("*")
```

```python
# Same number
n=int(input())
for i in range(1,n+1):
    for j in range(n):
        print(i,end="")
    print()
```

```python
# reverse same number
n=int(input())
for i in range(n,0,-1):
    for j in range(n):
        print(i,end="")
    print()
```

```python
# Same row
n=int(input())
for i in range(1,n+1):
    for j in range(1,i+1):
        print(j,end="")
    print()
```

```python
# reverse Same row
n=int(input())
for i in range(n,0,-1):
    for j in range(1,i+1):
        print(j,end="")
    print()
```

```python
# reverse Same row
n=int(input())
for i in range(n,0,-1):
    for j in range(n,n-i,-1):
        print(j,end="")
    print()
```

```python
# Number Pattern 
n=int(input())
for i in range(1,n+1):
    for s in range(n - i):
        print(" ", end="")
    for j in range(1,i+1):
        print(j,end="")
    for k in range(i-1,0,-1):
        print(k,end="")
    print()
```

```python
n = int(input())
i = 1
while i <= n:
    j = 1
    while j <= n - i:
        print(' ',end='')
        j+=1
    k = 1
    while k <= i:
        print(k,end='')
        k+=1
    l = i - 1
    while l >= 1:
        print(l,end='')
        l = l - 1
    i += 1
    print()
```
