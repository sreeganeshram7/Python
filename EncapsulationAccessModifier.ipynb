```python
# Encapsulation --->Encapsulation is the process of wrapping data and methods into a single unit and 
                 # restricting direct access to data by using controlled mechanisms and it is also called as data hiding
# How does encapsulation acheived in python? ----> by using access modifiers and @property
```

```python
# Access Modifier - It is used to define the level of access to the class variables and 
# methods controlling where they can be accessed form and they are used to implement encapsulation and data hiding
# In python access modifiers does not force strictly but use naming conventions
```

```python
# Types of access modifiers : public,private,protected
# public ---Members declared as public can be accessed from anywhere. Variable method()
class Student:
    def __init__(self,name):
        self.name=name
    def show(self):
        print(self.name)
s=Student("Ganesh")
s.show()
print(s.name)
```

```python
# protected ---Members declared as protected can access within the class and its child classes 
# It is indicated by using single underscore(_)                _variable,_method()
class Parent:
    def __init__(self):
        self._salary=50000 # protected variable
class Child(Parent):
    def show(self):
        print(self._salary)
c=Child()
c.show()   # It can be access in child class 
# Protected is a convention,not a strict protection
```

```python
# private ---Members declared as private can access only inside a class  
# It is indicated using double(__) underscore            __variable,__method()
class Parent:
    def __init__(self):
        self.__salary=60000 # private variable
    def show(self):
        print(self.__salary)
p=Parent()
p.show()
# p.__salary -- It shows the error because the __salary is not having in the class
```

```python
# Normal Getter and Setter----> method calls
class Student:
    def __init__(self,marks):
        self._marks=marks
    def get_marks(self):
        return self._marks
    def set_marks(self,value):
        if value<0 and value>100:
            raise ValueError("Invalid Marks")
            self._marks=value      
```

```python
s = Student(80)
s.set_marks(90) # It looks like function logic
print(s.get_marks())
```

```python
# @property getter and setter----> method-style access
class Student:
    def __init__(self,marks):
        self._marks=marks
    @property
    def marks(self):
        return self._marks
    @marks.setter
    def marks(self,value):
        if value<0 and value>100:
            raise ValueError("Invalid Marks")
            self._marks=value      
```

```python
s = Student(80)
s.marks = 90   # Normal
print(s.marks)
```

```python
# How to apply the validation isinstance() in Encapsulation
# It keeps data private 
# Access/modify the data only through methods or properties 
# Validate data before setting it
# isinstance() is used to check the datatype during validation
# Syntax: isinstance(value,datatype)
```

```python
# Encapsulation validation of getter and setter
class Student:
    def __init__(self,age):
        self.__age=None
        self.set_age(age)
   # setter with validation
    def set_age(self,age):
        if not isinstance(age,int):
            print("Age must be an integer")
        elif age<0:
            print("Age must be positive")
        else:
            self.__age=age
    # getter with validation
    def get_age(self):
        return self.__age
    
```

```python
s=Student(21)
s.get_age()
```

```python
s.set_age("twenty")
```

```python
s.set_age(-5)
```

```python
# Encapsulation validation @property
class Employee:
    def __init__(self,salary):
        self.__salary=None
        self.salary=salary  # calls setter
    @property
    def salary(self):
        return self.__salary
    @salary.setter
    def salary(self,value):
        if not isinstance(value,(int,float)):
            raise TypeError("Salary must be a number")
        if value<=0:
            raise ValueError("Salary must be greater than zero")
            self.__salary=value
```

```python
e = Employee(50000)
print(e.salary)      
```

```python
e.salary = 60000   
print(e.salary)
```

```python
e.salary=-60000
print(e.salary)
```

```python
e.salary="abc"
print(e.salary)
```
