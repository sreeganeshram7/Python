Section D – Time‑based analysis
Convert Date to proper datetime and extract:

Year

Month

Day of week

For each month, compute:

Total revenue

Number of transactions

Average TotalAmount per transaction

On which day of the week is revenue highest on average?

Identify any months where both:

Total revenue is above the overall monthly average

Number of transactions is also above the overall monthly average


```python
df['Year']=df['Date'].dt.year
df       
```

```python
df['Month']=df['Date'].dt.month
df   
```

```python
df['DayOfWeek']=df['Date'].dt.day_name()
df
```

```python
monthly=df.groupby('Month').agg(TotalRevenue=('TotalAmount','sum'),NumberofTransactions=('TransactionID','count'),AverageRevenue=('TotalAmount','mean'))
```

```python
monthly
```

```python
top_revenue=df.groupby('DayOfWeek')['TotalAmount'].mean().sort_values(ascending=False).idxmax()
top_revenue
```

```python
avg_revenue = monthly['TotalRevenue'].mean()
avg_revenue
```

```python
avg_transactions = monthly['NumberofTransactions'].mean()
avg_transactions
```

```python
filtered_month=monthly[(monthly['TotalRevenue']>avg_revenue)|( monthly['NumberofTransactions']>avg_transactions)]
filtered_month
```
