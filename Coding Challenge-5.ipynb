Question 1:

Define a python function to print the longest word in the list of technologies given below.

myList = ['Java', 'Data Science', 'Python', 'Machine Learning', 'Artificial Intelligence']

```python
def longest_word(lst):
    longest = lst[0]
    
    for word in lst:
        if len(word) > len(longest):
            longest = word
    
    print("Longest word:", longest)
```

```python
longest_word(['Java', 'Data Science', 'Python', 'Machine Learning', 'Artificial Intelligence'])
```

```python
def longest_word(lst):
    return max(lst, key=len)
```

```python
longest_word(['Java', 'Data Science', 'Python', 'Machine Learning', 'Artificial Intelligence'])
```

```python
import sys

def nextPalindrome(n):
    numStr = str(n)
    for i in range(n + 1, sys.maxsize):
        if str(i) == str(i)[::-1]:            
            return i
```

### Question 4:

Write a python function to check if the list of numbers given below is in Arithmetic Progression or not.

myList = [5, 7, 9, 11]



```python
def is_ap(lst):
    if len(lst)<2:
        return True
    diff=lst[1]-lst[0]
    for i in range(2,len(lst)):
        if lst[i]-lst[i-1]!=diff:
            return False
        return True
```

```python
is_ap([5,7,9,11])
```

```python
def is_ap(lst):
    return all(lst[i] - lst[i-1] == lst[1] - lst[0] for i in range(2, len(lst)))
```

```python
is_ap([5,7,9,11])
```

### Question 5:

Write a python class vowelsCheck to find the number of vowels in a string. Take the string input from the user.



```python
class vowelsCheck:
    def __init__(self, myStr):        
        self.text = myStr
        self.count = 0

    def vowel_count(self):
        for i in self.text:
            if i in ("a","e","i","o","u"):
                self.count += 1
        print("Number of vowels: ",self.count)
```

```python
s = input("Enter a string: ")
myObj = vowelsCheck(s)
myObj.vowel_count()
```

### Question 6:

Write a python exception handling program to check if the date of February entered is a multiple of 7 or not. Print the message 'Incorrect Date' if the date exceeds 28. For all other exceptions, print the message 'Invalid Data.' (Hint: User-defined exception)

#### Sample Input - 1:

Enter the date of February: 30

#### Expected Output:

Incorrect Date.

#### Sample Input - 2:

Enter the date of February: aa

#### Expected Output:

Invalid Data.

#### Sample Input - 3:

Enter the date of February: 21

#### Expected Output:

'Multiple of 7.'

```python
class InvalidDateError(Exception):
    pass

def date_check(input_date):
    while True:
        try:
            input_date = int(input('Enter the date of February: '))
            if input_date <= 28:
                if input_date % 7 == 0:
                    return 'Multiple of 7.'
                else:
                    return 'Not a multiple of 7.'
            else:
                raise InvalidDateError

        except InvalidDateError:
            print('Incorrect Date.')
        except:
            print('Invalid Data.')
```

```python
date_check(30)
```

### Question 9:

We have a bag of 9 balls, which has a discrete number from 1 to 9 marked on each of them. But when we take out all the balls from the bag, few of them are missing. Given below is the list of balls that are present.

myList = [1, 2, 3, 5, 6, 7, 9]

Write a python function to find the missing balls.

#### Expected Output:

[4, 8]

```python
def find_missing(lst):
    m=[]
    for i in range(1,10):
        if i not in lst:
            m.append(i)
    return m
```

```python
find_missing([1,2,3,5,6,7,9])
```

### Question 10:

Create a class Fan which is initialized with its brand. The class contains two instance methods for switching on the fan with speed five and switching off the fan with speed 0. Create another method to show the current status of the fan.

#### Sample Input:

Brand = MONTE CARLO
<br>Scene - 1: Switch-on the fan and show the status
<br>Scene - 2: Switch-off the fan and show the status

#### Expected Output:

The fan manufactured by MONTE CARLO is currently Running with a speed of 5
<br>The fan manufactured by MONTE CARLO is currently Idle with a speed of 0

```python
class Fan:
    def __init__(self, brand):
        self.brand = brand        
        self.speed = 0
        self.state = 'Idle'    

    def switch_on(self):
        self.state = 'Running'
        self.speed = 5

    def switch_off(self):
        self.state = 'Idle'
        self.speed = 0
        
    def show_status(self):
        print(f'The fan manufactured by {self.brand} is currently {self.state} with a speed of {self.speed}')

fan_obj = Fan('MONTE CARLO')
fan_obj.switch_on()
fan_obj.show_status()
fan_obj.switch_off()
fan_obj.show_status()
```
