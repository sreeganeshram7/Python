### Write a function that takes a tuple of integers and returns a new tuple containing only those elements which appear exactly once, preserving the original order.

```python
def unique_once(value):
    count={}
    result=[]
    for i in value:
        count[i]=count.get(i,0) +1

    for j in value:
        if count[j]==1:
            result.append(j)

    return tuple(result)
```

```python
unique_once([1,2,3,2,2,4,4,5])
```

### Given a tuple t, return a new tuple where all elements are rotated right by k positions (negative k means left rotation). Do not convert t to a list.

```python
#rotation of tuple by 1 digit 4123
#                     2       3412
#                     3        2341
```

```python
def rotated_tuple(t,k):
    l=len(t)
    k=k%l
    return t[-k:]+t[:-k]
print (rotated_tuple((1,2,3,4),2))
```

### You are given a nested tuple like ((1, 2), (3, (4, 5)), 6, ((7,), 8)). Write a recursive function to flatten it into a single-level tuple of integers.

```python
result=[]
def recurse(item):
        if isinstance(item, tuple):
            for sub in item:
                recurse(sub)
        else:
            result.append(item)
        return tuple(result)
print(recurse(((1,2),(3,(4,5)),6)))
```
