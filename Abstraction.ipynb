```python
# how is abstraction is acheived i python?
# 1.Abstract classes(abc module)
# 2.Abstract methods(@abstractmethod decorator)
```

```python
# Abstract class:It cannot be instantiated and it can contain abstract and normal methods
# Absstract methods: It has no implementation in parent class and it must be implemented by child classes
```

```python
# Main purpose of ABC -Abstract base classes-it is used to define the common structure (blueprint) for related classes
# 1.Enforces method implementation 2.it defines a common interfaces 3.it prevents incomplete classes 4.it supports abstraction and polymorphism
```

```python
from abc import ABC ,abstractmethod
class Shape(ABC):
    @abstractmethod
    def area(self):
        pass
class Rectangle(Shape):
    def area(self):
        return 10*5
r=Rectangle()
r.area()
```

```python
class Circle(Shape):
    def area(self):
        return 3.14*7*7
c=Circle()
c.area()
```

```python
from abc import ABC , abstractmethod
class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass
class Dog(Animal):
    def sound(self):
        print("Dog barks")
```

```python
d=Dog()
d.sound()
```

```python
from abc import ABC, abstractmethod
class School(ABC):
    @abstractmethod
    def board(self):
        pass
class Teacher(School):
    def board(self):
        print("It is using to write")
```
